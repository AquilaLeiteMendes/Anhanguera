```python
# Sistemas de Gestão de Notas de Alunos
def adicionar_notas():
    notas = []
    print("--- Sistema de Gestão de Notas ---")
    while True:
        entrada = input("Digite a nota do aluno (ou 'F' para finalizar): ")
        if entrada.upper() == 'F':
            break
        try:
            nota = float(entrada)
            if 0 <= nota <= 10:
                notas.append(nota)
            else:
                print("Por favor, insira uma nota entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Digite um número.")
    return notas

# Função para cálculo da média
def calcular_media(notas):
    if not notas:
        return 0
    return sum(notas) / len(notas)

# Função para determinação da situação (Aprovado >= 7, Reprovado < 7)
def determinar_situacao(media):
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"

# Função para exibir o relatório final
def exibir_relatorio(notas, media, situacao):
    print("\n--- Relatório Final ---")
    print(f"Notas Inseridas: {notas}")
    print(f"Média Final: {media:.2f}")
    print(f"Situação do Aluno: {situacao}")

# Execução principal do programa com tratamento para lista de notas vazia
try:
    notas_aluno = adicionar_notas()
    if len(notas_aluno) == 0:
        raise ValueError("Erro: Nenhuma nota foi registrada. Não é possível gerar o relatório.")

    media_aluno = calcular_media(notas_aluno)
    situacao_aluno = determinar_situacao(media_aluno)
    exibir_relatorio(notas_aluno, media_aluno, situacao_aluno)
except ValueError as e:
    print(f"\n{e}")
```

```python
# ============================================================
# Sistema de Gestão de Notas de Alunos
# Disciplina: Linguagem de Programação
# Unidade: U1_INTRODUÇÃO_A_LINGUAGEM_PYTHON
# Aula: A4_FUNÇÕES_EM_PYTHON
# ============================================================

def adicionar_notas():
    """
    Função responsável por cadastrar as notas do aluno.
    Utiliza uma lista para armazenar as notas e um loop
    de repetição para permitir a inserção de quantas notas
    o usuário desejar. A entrada é interrompida quando o
    usuário digita 'sair'.
    """
    notas = []  # Lista vazia que irá armazenar as notas
    print("=" * 50)
    print("CADASTRO DE NOTAS")
    print("Digite as notas uma a uma.")
    print("Para finalizar o cadastro, digite 'sair'.")
    print("=" * 50)

    while True:  # Estrutura de repetição
        entrada = input("Informe a nota (ou 'sair'): ").strip()

        if entrada.lower() == "sair":
            break  # Encerra o loop quando o usuário deseja parar

        try:
            # Aceita tanto ponto quanto vírgula como separador decimal
            nota = float(entrada.replace(",", "."))
            if 0 <= nota <= 10:
                notas.append(nota)  # Adiciona a nota válida na lista
                print(f"Nota {nota} adicionada com sucesso!")
            else:
                print("Atenção: a nota deve estar entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Digite um número ou 'sair'.")

    return notas


def calcular_media(notas):
    """
    Função que calcula a média aritmética das notas.
    Recebe a lista de notas como parâmetro e retorna a média.
    Se a lista estiver vazia, retorna 0 para evitar divisão por zero.
    """
    if len(notas) == 0:
        return 0.0
    media = sum(notas) / len(notas)
    return media


def determinar_situacao(media):
    """
    Função que determina a situação do aluno com base na média.
    Utiliza estrutura condicional (if/else):
    - Média >= 7 → Aprovado
    - Média < 7  → Reprovado
    """
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"


def exibir_relatorio(notas, media, situacao):
    """
    Função que exibe o relatório final contendo:
    - Todas as notas inseridas
    - A média calculada
    - A situação do aluno (Aprovado/Reprovado)
    """
    print("\n" + "=" * 50)
    print("RELATÓRIO FINAL")
    print("=" * 50)
    print(f"Notas Cadastradas: {notas}")
    print(f"Média do Aluno:   {media:.2f}")
    print(f"Situação:         {situacao}")
    print("=" * 50)

# Execução principal do programa com tratamento para lista de notas vazia
try:
    notas_aluno = adicionar_notas()
    if len(notas_aluno) == 0:
        raise ValueError("Erro: Nenhuma nota foi registrada. Não é possível gerar o relatório.")

    media_aluno = calcular_media(notas_aluno)
    situacao_aluno = determinar_situacao(media_aluno)
    exibir_relatorio(notas_aluno, media_aluno, situacao_aluno)
except ValueError as e:
    print(f"\n{e}")
```
